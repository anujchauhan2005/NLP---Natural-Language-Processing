```python
nlp = spacy.load('en_core_web_sm')
```

```python
text = nlp('GFG is looking for Data Science Interns')
text
```

## 1. Tokenization

```python
for token in text:
  print(token)
```

```python
for token in text:
  print(token.text)
```

```python
text = nlp('The cost of Iphone in U.K is 699$') # here $ is consider as single token
```

```python
for token in text:
  print(token)
```

## 2. Parts of Speech(POS)

---



```python
for token in text:
  print(token.text, token.pos)
```

```python
text
```

```python
for token in text:
  print(token.text, token.pos_)    # this prints the parts of speech noun,pronoun,adjective
  # here we get what each word represents
```

## 3. Sentence Tokenization

```python
text = nlp('This is sentence one. This is second sentence. This is Last one. Lets study now')
```

```python
for sent in text.sents:
  print(sent)
```

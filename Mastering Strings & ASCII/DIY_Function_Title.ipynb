```python
def convert_upper(txt):
    string = ''

    for char in txt:
        if ord(char) >= 97 and ord(char) <= 122:    # 97 - 122 (Lowercase)
            string += chr(ord(char) - 32)           # Subtract 32 to make it Uppercase
        else:
            string += char

    return string


def convert_lower(txt):
    string = ''

    for char in txt:
        if ord(char) >= 65 and ord(char) <= 90:         # 65 - 90 (uppercase)
            string += chr(ord(char) - 32)                # Add 32 to make it Lowercase
        else:
            string += char

    return string
```

```python
def title(txt):
   output = []

   for word in txt.split(' '):
       output.append(convert_upper(word[0]) +convert_lower(word[1:]))

   return ' '.join(output)
```

```python
title('anuj pratap singh chauhan')
```

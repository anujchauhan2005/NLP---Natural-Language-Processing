## 1. Uppercase Function (65 - 90)

```python
# 65 - 90  (Uppercase)   words limit

# 97 - 122 (LowerCase)    words limit
```

```python
def convert_upper(txt):
    string = ''

    for char in txt:
        if ord(char) >= 97 and ord(char) <= 122:    # 97 - 122 (Lowercase)
            string += chr(ord(char) - 32)           # Subtract 32 to make it Uppercase
        else:
            string += char

    return string
```

```python
convert_upper(txt)
```

```python
txt.upper()
```

## 2. Lowercase Function

```python
def convert_lower(txt):
    string = ''

    for char in txt:
        if ord(char) >= 65 and ord(char) <= 90:         # 65 - 90 (uppercase)
            string += chr(ord(char) - 32)                # Add 32 to make it Lowercase
        else:
            string += char

    return string
```

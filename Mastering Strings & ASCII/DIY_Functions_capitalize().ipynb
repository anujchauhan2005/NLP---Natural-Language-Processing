```python
text = 'the quick brown fox jumps over the LAZY dog'
```

```python
def capitalize(text):
  if (len(text)!=0):
    return text[0].upper() + text[1:].lower()

  else:
    return None



```

```python
capitalize('Hello,how are You')
```

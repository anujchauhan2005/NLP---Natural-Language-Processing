## Text to Sequences

```python


from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tok = Tokenizer()
```

```python
corp = ['coffe is hot','water is cold']

tok.fit_on_texts(corp)

tok.word_index
```

```python
tok.texts_to_sequences(corp)   # gives sentence number
```

```python
tok.texts_to_sequences(['Water is hot','black coffe is cold'])
```

## Adding OVV

```python
from tensorflow.keras.preprocessing.text import Tokenizer

tok = Tokenizer(oov_token = 'black')
corp = ['coffe is hot','water is cold']

tok.fit_on_texts(corp)
print(tok.word_index)

tok.texts_to_sequences(['Water is hot','black coffe is cold'])
print(tok.texts_to_sequences(corp))
```

```python
tok.word_index
```

## Limiting the number of words

```python
from tensorflow.keras.preprocessing.text import Tokenizer

tok = Tokenizer( num_words = 4)
corp = ['coffe is hot','water is cold']

tok.fit_on_texts(corp)
print(tok.word_index)

tok.texts_to_sequences(['Water is hot','black coffe is cold'])
```

## Using NLTK

```python
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
word_tokenize(txt)
```

```python
sent_tokenize(txt)
```

```python
for word in word_tokenize(txt):
  if(word != '.'):      # this lines remove (.) word
    print(word)
```

```python
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize

corpus = "India is a land of unity and diversity. It is the seventh-largest country and the second most populated in the world. India has multiple religions, languages, and cultures, but people live together peacefully. From historical monuments like the India Gate to colourful festivals such as Holi and Diwali, the country embraces every tradition. Indian farmers work hard to feed the nation. The Indian Army protects us from every danger, while the Indian Constitution teaches us about rights and responsibilities. Schools celebrate India’s achievements on Independence Day and Republic Day. India’s values encourage us to respect others, help the needy, and work for the country’s progress. This paragraph on India inspires students to love their nation and contribute to its bright future."
corpus
```

## StopWords Removal

```python
words = []

for word in word_tokenize(corpus):
  if ((word.lower() not in stopwords.words('english')) and len(word) >= 2):   # this gives unique words
         words.append(word.lower())
```

```python
words
```

## Making Vocab

```python
vocab = list(set(words))
```

## 2. Findall

```python
a = 'Hello,how are you?'
b = 'are'    # only same spelling or words are matched uppercase amnmnd lowecase matters

re.findall(b,a)
```

```python
a = 'Hello,how are you?'
b = 'are'    # only same spelling or words are matched uppercase amnmnd lowecase matters

len(re.findall(b,a))    # prints how many times it was  printed in semntence
```

## 3. Finditer

```python
for s in re.finditer('are','Hello, how are area are you?'):
  print(s.span())
```

## 4. Creating Regular Expressions

```python

txt = 'My telephone number is  9399892005'
pattern = '\d\d\d\d\d\d\d\d\d\d'   # 10 digit

re.search(pattern,txt)
```

```python

txt = 'My telephone number is  9399892005'
pattern = '\d\d\d\d\d\d\d\d\d'   # 9 digit

re.search(pattern,txt)   # it matches only the first 9 digit
```

```python

txt = 'My telephone number is  939-9892-005'
pattern = '\d\d\d-\d\d\d\d-\d\d\d'   # 10 digit

re.search(pattern,txt)
```

```python

txt = 'My telephone number is  939-9892-005'
pattern = '\d{3}-\d{4}-\d{3}'   # this gives exactly same answer

re.search(pattern,txt)
```

```python

txt = 'My telephone number is  939-9892-005'
pattern = '\d{3}-\d{4}-\d{3}'   # this gives exactly same answer

re.search(pattern,txt).group()
```

```python

txt = 'My telephone number is  9399892005'
pattern = '\w\w\w\w\w'   # 5 words

re.search(pattern,txt)
```

```python
re.search(pattern,txt).group()
```

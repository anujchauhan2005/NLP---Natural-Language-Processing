```python
text = 'The quick brown for jumps over the lazy dog'
```

## 1. Strings to ASCII Conversion

```python
ascii_ = [{char: ord(char)} for char in text]
ascii_
```

## 2. ASCII to string Conversion

```python
ascii_ = [ord(char) for char in text]
```

```python
ascii_
```

```python
''.join([chr(asc) for asc in ascii_])    # covert ascii codes back to strings
```

```python
import re

```

## 1. Finding pattern with specific ending

```python
re.findall('at','The rat sat on the mat and attached by a cat')
```

```python
re.findall('.at','The rat sat on the mat and attached by a cat')   # in this only 3 words are there ending with _at
```

```python
re.findall('..at','The rat sat on the mat and attached by a cat')
#in this space is also printed
# in this only 4 words are there ending with __at
```

## 2. Start and End Pattern Matching

```python
re.findall('\d','This is a number 64335')    # here \d represents digits
```

```python
re.findall('\d','This is a number 6 not 4')
```

```python
re.findall('\d','4 is divisible by 2 not by 3')
```

```python
re.findall('\d$','4 is divisible by 2 not by 3')
 # here $ prints the last digit if there is digit in the last then it print if not digit then it retuns empty list
```

```python
re.findall('^\d','4 is divisible by 2 not by 3')
# nhere ^ represents the begining if there is a digit in the begining then it prints those number if not then it prints empty list even space is also counted
```

## 3. Finding Capital Letters

```python
re.findall('[A-Z]','Welcome to GFG')   # this prints only the uppercase letters
```

## 4. Finding the small Letters

```python
re.findall('[a-z]','Welcome to GFG')   # this prints only the small letters lowercase
```

## 5. Finding the Uppercase and Lowercase

```python
re.findall('[A-Za-z]','Welcome to GFG')   # here space is not included
```

## 6. Finding Uppercase,Lowercase and Numbers

```python
re.findall('[A-Za-z0-9]','Welcome to GFG 1')    # here all are mixed and space is not included
```

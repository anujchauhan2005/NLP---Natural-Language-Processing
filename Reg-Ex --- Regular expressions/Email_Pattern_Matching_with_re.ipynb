```python
import re
mail = 'anujchauhan9399@itm.ac.in'
p = '[A-Za-z0-9]+@[\w]+.[\w]+'

re.search(p,mail)
```

```python
re.search(p,'anujchauhan9399@itm.ac.in  dsadasfsafads df').group()
```

```python
re.search(p,'9876@itm.com')   # without using @ it is not recognised
```

## Email ID With Specific Domain

```python
mail = 'ashish@gfg.org'

p = '[A-Za-z0-9]+@(gfg).(org)$'

re.search(p,mail)
```

```python
re.search(p,'ashish@gfg.in')  # not matching thats why it does not return anything
```

```python
re.search(p,'prakash@gfg.org')   # matching
```

```python
re.search(p,'ashish.jangra@gfg.org')
```

## Email ID With Multiple Selected Domain

```python
mail = 'ashish@gfg.org'

p = '[A-Za-z0-9]+@[A-Za-z]+.(org|edu|gov|net|in)'   # these checks in the lasts words after(.)

re.search(p,mail)
```

```python
re.search(p,'ashish@gfg.com')   #not matching
```

```python
re.search(p,'ashish@gfg.edu')  # matching
```

```python
re.search(p,'ashish@gfg.gov')  # matching
```

```python
re.search(p,'ashish@gfg.in')
```

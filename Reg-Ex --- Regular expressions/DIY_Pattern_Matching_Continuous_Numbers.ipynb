## 1. Continous Numbers

```python
txt = 'The Number is 8372493 not 423423'

c = 0
index  = []

for i in range(len(txt)):
    if (txt[i].isdigit() == True):  # check if digit

      if(c == 0):    # count no of digit
         st = i      # starting the no
      c += 1          # incrementing the counter for that number


    else:             # if not digit
      if(c != 0):      # if coming from a digit
        end = i       # ending the number

        print(txt[st:end])    # printing the number
      c = 0


if (c !=0):    # if the number is at the end of the string
  print(txt[-c:])

```

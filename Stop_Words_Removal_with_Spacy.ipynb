## 5. Removing StopWords from corpus

```python
text = '''Data science is one of the most important fields in modern technology.\nIt helps organizations collect, clean, analyze, and understand large amounts of data.\nData science combines statistics, mathematics, programming, machine learning, artificial intelligence, and data visualization to find useful patterns and insights.\nA data scientist works with raw data and uses different techniques to prepare the data for analysis.\nThe data is often collected from different sources and may contain missing values, duplicate records, errors, and unnecessary information.\nAfter cleaning the data, data scientists perform exploratory data analysis to understand the patterns and relationships in the dataset.\nMachine learning algorithms are then used to build models that can make predictions and support better decisions.\nData visualization helps people understand complex data through graphs, charts, and dashboards.\nData science is used in many industries such as healthcare, finance, banking, e-commerce, marketing, education, and transportation.\nIt can help companies understand their customers, predict future trends, detect fraud, improve products, and make better business decisions.\nTherefore, data science is an important field that converts raw data into meaningful information and helps organizations make data-driven decisions.'''

text = text.replace('\n','')
text = text.replace('  ','')
text = text.strip()

text
```

```python
corp = nlp(text)
corp
```

## 5.1 Findings rhe stopwords from the corpus

```python
stop_word = []

for token in corp:
  if token.is_stop:
    stop_word.append(token.text)
print(stop_word)
print(len(stop_word))
```

## 5.2 Finding the words that dosen't belong to stop_words

```python
for token in corp:
  if not token.is_stop:
    print(token)

```

```python
' '.join([token.text  for token in corp if not token.is_stop ])
```

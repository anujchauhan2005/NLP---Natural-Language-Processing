```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt')


from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer, PorterStemmer    # for stemming
```

```python
stem = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

## Lemmatization

```python
print(lemmatizer.lemmatize('change'))     # these all gives root words of that word which make sense
print(lemmatizer.lemmatize('changes'))
print(lemmatizer.lemmatize('changer'))
print(lemmatizer.lemmatize('changed'))
```

## Stemming

```python
print(stem.stem('change'))    # This gives the root word which does not make any sense
print(stem.stem('changes'))
print(stem.stem('changer'))
print(stem.stem('changed'))
```

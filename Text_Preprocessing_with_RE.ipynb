```python
import re
```

## 1. Removing the special characters


```python
txt = 'Eshant is happy! Bu@t, wh#y is he no&t p*laying?'
```

```python
' '.join(re.findall('[^!@#&*?]+',txt))   # this removes the special characters
```

## 2. Exclusion

```python
txt = "I'm 24"

''.join(re.findall('[\d\d]',txt))
```

```python
# prints all the things expects digits
txt = "I'm  Anuj Chauhan and 24"

print(''.join(re.findall('\D',txt)))    # we use \D instead of \d for the other things expect digits
```

## 3. Finding Patterns Link Geeks-For-Geeks

```python
txt = "Hello I am studing from Geeks-for-Geeks and it is amazing. Let's see how this work-it-out"
```

```python
re.findall('\w\w\w\w',txt)
```

```python
re.findall('[\w]+-[\w]+-[\w]+',txt)
```

## 4. Phone No. Pattern Matching

```python
for no in [ "653-5556-6543" , "888-9999-7574", " 757-3345-455"]:
  print(re.findall('[\d]+-[\d]+-[\d]',no))
  print(str(re.findall('[\d]+-[\d]+-[\d]',no)[0]).replace('-',''))
```
